```python
print(bank_clean["Personal_Loan"].value_counts())
```

```python
print(
    bank_clean["Personal_Loan"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
print("Average income:",
      bank_clean["Income"].mean())
```

```python
print("Income of customers who rejected loan:",
      bank_clean.loc[
          bank_clean["Personal_Loan"] == 0, "Income"
      ].mean())
```

```python
print("Income of customers who accepted loan:",
      bank_clean.loc[
          bank_clean["Personal_Loan"] == 1, "Income"
      ].mean())
```

```python

print("Average age:", bank_clean["Age"].mean())
```

```python
print("Rejected loan:",
      bank_clean.loc[
          bank_clean["Personal_Loan"] == 0, "Age"
      ].mean())
```

```python
print("Accepted loan:",
      bank_clean.loc[
          bank_clean["Personal_Loan"] == 1, "Age"
      ].mean())
```

```python
print(
    bank_clean.groupby(
        ["Education", "Personal_Loan"]
    ).size().unstack(fill_value=0)
)
```
